## Q1.
Take the roll number, extract its digits, multiply each digit by 10, and perform the required list operations.

```python
roll_no = input("Enter your roll number: ").strip()
digits = [int(d) for d in roll_no if d.isdigit()]
L = [d * 10 for d in digits]

print("L =", L)

L.append(50)
print("After append(50):", L)

L.insert(2, 70)
print("After insert(2, 70):", L)

L.remove(50)
print("After remove(50):", L)

removed = L.pop()
print("Popped element:", removed)
print("After pop():", L)

L.sort()
print("Ascending:", L)
L.sort(reverse=True)
print("Descending:", L)

print("First three:", L[:3])
print("Last three:", L[-3:])

average = sum(L) / len(L)
above_average = [x for x in L if x > average]
print("Average:", average)
print("Elements greater than average:", above_average)
```

## Q2.
Create `scores` from the first 8 values of `L` and perform the required tuple operations.

```python
scores = tuple(L[:8])
print("scores =", scores)

highest = max(scores)
lowest = min(scores)
print("Highest score:", highest)
print("Index of highest:", scores.index(highest))
print("Lowest score:", lowest)
print("Lowest occurs:", scores.count(lowest), "time(s)")

# Tuples are immutable, so they cannot be reversed in place.
reversed_scores = list(reversed(scores))
print("Reversed tuple as list:", reversed_scores)

score = int(input("Enter a score to search: "))
if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score is not present.")

try:
    scores[0] = 100
except TypeError as e:
    print("Captured error:", e)
    print("Tuple elements cannot be changed because tuples are immutable.")

first_score, second_score, *remaining_scores = scores
print("First:", first_score)
print("Second:", second_score)
print("Remaining:", remaining_scores)
```

## Q3.
Use the roll number as the random seed, generate 100 numbers from 100 to 900, count odd/even/prime numbers, and find the most frequent number.

```python
import random
from collections import Counter

random.seed(int(roll_no))
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

odd_numbers = [x for x in numbers if x % 2 != 0]
even_numbers = [x for x in numbers if x % 2 == 0]
print("Odd count:", len(odd_numbers))
print("Even count:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

number, count = Counter(numbers).most_common(1)[0]
print("Most frequent number:", number)
print("Occurrences:", count)
```

## Q4.
Use the first 8 digits of the roll number to create sets `A` and `B`, then perform the required set operations.

```python
roll_digits = [int(d) for d in roll_no if d.isdigit()][:8]

A = {d * 7 for d in roll_digits}
B = {d * 9 for d in roll_digits}

print("A =", A)
print("B =", B)

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
print("Symmetric difference:", A.symmetric_difference(B))

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

x = int(input("Enter a value to discard from A: "))
A.discard(x)
print("A after discard:", A)
```
