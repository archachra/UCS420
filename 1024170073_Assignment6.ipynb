```python
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 500)

y1 = M * x**2
y2 = M * np.sin(x)

plt.plot(x, y1, 'r-', label='y = Mx²')
plt.plot(x, y2, 'b--', label='y = M sin(x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Mathematical Functions')
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Subject': ['Math', 'Physics', 'Chemistry', 'Computer', 'English'],
    'Score': [85, 78, 92, 95, 80]
}

df = pd.DataFrame(data)

sns.barplot(x='Subject', y='Score', data=df, hue='Subject', palette='Set2', legend=False)

for i, score in enumerate(df['Score']):
    plt.text(i, score + 1, str(score), ha='center')

plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.title('Scores in Five Subjects')
plt.grid(axis='y')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

roll_no = 1024170073
np.random.seed(roll_no)
data = np.random.randn(50)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(np.cumsum(data))
axes[0, 0].set_title('Cumulative Sum')
axes[0, 0].set_xlabel('Index')
axes[0, 0].set_ylabel('Cumulative Sum')
axes[0, 0].grid(True)

x = np.arange(50)
noise = np.random.randn(50)

axes[0, 1].scatter(x, data + noise)
axes[0, 1].set_title('Scatter Plot with Random Noise')
axes[0, 1].set_xlabel('Index')
axes[0, 1].set_ylabel('Value')
axes[0, 1].grid(True)

axes[1, 0].plot(data)
axes[1, 0].set_title('Random Data')
axes[1, 0].set_xlabel('Index')
axes[1, 0].set_ylabel('Value')
axes[1, 0].grid(True)

axes[1, 1].hist(data, bins=10)
axes[1, 1].set_title('Distribution of Random Data')
axes[1, 1].set_xlabel('Value')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"
df = pd.read_csv(url)
```

```python
# 1
sns.lineplot(x='month_number', y='total_profit', data=df, marker='o')

plt.title('Total Profit of All Months')
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.grid(True)
plt.show()
```

```python
# 2
products = [
    'facecream',
    'facewash',
    'toothpaste',
    'bathingsoap',
    'shampoo',
    'moisturizer'
]

for product in products:
    sns.lineplot(
        x='month_number',
        y=product,
        data=df,
        marker='o',
        label=product
    )

plt.title('Product Sales Data')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 3
df_melted = df.melt(
    id_vars='month_number',
    value_vars=products,
    var_name='Product',
    value_name='Sales'
)

sns.barplot(
    x='month_number',
    y='Sales',
    hue='Product',
    data=df_melted
)

plt.title('Monthly Sales of All Products')
plt.xlabel('Month Number')
plt.ylabel('Units Sold')
plt.legend(title='Product')
plt.grid(axis='y')
plt.show()
```
