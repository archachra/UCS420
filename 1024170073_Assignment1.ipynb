```python
n = int(input("Enter n: "))

sum = 0

for i in range(2, n + 1):
    is_prime = True

    for j in range(2, i):
        if i % j == 0:
            is_prime = False
            break

    if is_prime:
        sum += i

print(sum)
```

6.2 WAP USING FUNCTION THAT ADD ALL PRIME NUMBERS FROM 1 TO N,GIVEN BY THE USER

```python
def is_prime(n):
  for i in range(1,n+1):
    if n%i==0 and i!=1 and i!=n:
      return False
  return True
n=int(input("enter a number: "))
sum=0
for i in range(2,n+1):
  if is_prime(i):
    sum+=i
print(sum)
```
