**Q1 Build Your Personalized Knowledge Base**

```python
import pandas as pd

fixed_entries = [
{"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",
"keywords": "fee cost price charge", "category": "billing"},
{"question": "how to reset password", "answer": "Go to Settings > Reset Password.",
"keywords": "password reset login", "category": "account"},
{"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",
"keywords": "hours timing open time", "category": "general"},
{"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.",
"keywords": "pay payment upi fee", "category": "billing"},
]
```

```python
roll_no = 1024170073
last_two_digits = [int(d) for d in str(roll_no)[-2:]]
categories = ["billing", "account", "general"]

custom_entries = []
for d in last_two_digits:

    category = categories[d % 3]

    if category == "billing":
        question = "how can i get a refund for my payment"
        answer = "You can request a refund through the billing section."
        keywords = "refund payment billing"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number update"

    else:
        question = "how can i change my account details"
        answer = "You can change your details from the account settings."
        keywords = "account details update"

    custom_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })

all_entries = fixed_entries + custom_entries
df = pd.DataFrame(all_entries)
print(df)
```

**Q2 Generate and Score a Hypothesis**

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []
    for _, row in df.iterrows():

        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(text.split())
        matched_words = query_words.intersection(entry_words)

        if len(query_words) > 0:
            confidence = len(matched_words) / len(query_words)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })

    results.sort(key=lambda x: x["confidence"], reverse=True)

    return results


query = "how do I reset my password"
matches = score_query(query, df)
for result in matches:
    print(result)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(text.split())
        matched_words = query_words.intersection(entry_words)

        if len(query_words) > 0:
            confidence = len(matched_words) / len(query_words)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })

    results.sort(key=lambda x: x["confidence"], reverse=True)

    if results:
        highest_score = results[0]["confidence"]
        top_matches = [result for result in results if result["confidence"] == highest_score]

        print("Top matching entries:")
        for result in top_matches:
            print(result)

    return results


print("Tie Query")
score_query("fee", df)
print("\nNon-Tie Query")
score_query("reset password", df)
```
