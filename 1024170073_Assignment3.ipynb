**Q.6 Create a sample dataset (employees.csv) containing information about employees in
a company.**


```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
```

```python
# a)
print("Shape:")
df.shape
```

```python
# b)
print("\nDataFrame Information:")
df.info()
```

```python
# c)
print("\nDescriptive Statistics:")
df.describe()
```

```python
# d)
print("\nFirst 5 rows:")
df.head(5)
```

```python
print("\nLast 3 rows:")
df.tail(3)
```

```python
# e)
print("Average Salary:", df["Salary"].mean())
print("Total Bonus:", df["Bonus"].sum())
print("Youngest Age:", df["Age"].min())
print("Highest Rating:", df["Rating"].max())
```

```python
# f)
df = df.sort_values("Salary", ascending=False)
print("\nSorted by Salary:")
df
```

```python
# g)
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
df["Performance"] = df["Rating"].apply(performance)
```

```python
# h)
print("Missing Values:")
df.isnull().sum()
```

```python
# i)
df = df.rename(columns={"Employee_ID": "ID"})
```

```python
# j)
print("Employees with more than 5 years experience:")
df[df["Years_of_Experience"] > 5]
```

```python
print("\nEmployees in IT department:")
df[df["Department"] == "IT"]
```

```python
# k)
df["Tax"] = df["Salary"] * 0.10
print("Final DataFrame:")
df
```

```python
# l)
df.to_csv("modified_employees.csv", index=False)
df
```
