```python
import math
n = int(input("Enter a number: "))
sum = 0
for i in range(2, n + 1):
    prime = True
    for j in range(2, int(math.sqrt(i)) + 1):
        if i % j == 0:
            prime = False
            break
    if prime:
        sum += i

print(sum)
```

```python
def addAllNum(n):
  sum=0
  for i in range(1,n):
    if(i%2!=0):
      sum=sum+i
  return sum
n=int(input("number?"))
print(addAllNum(n))
```

```python
def addAllPrime(n):
  sum = 0
  for i in range(2, n + 1):
      prime = True
      for j in range(2, int(math.sqrt(i)) + 1):
          if i % j == 0:
              prime = False
              break
      if prime:
          sum += i

  return(sum)
n=int(input("number?"))
print(addAllPrime(n))
```
